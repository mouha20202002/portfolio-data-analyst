# **Importation du jeu de données** #

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# **Chargement du jeu de données** #

```python
# Charger le jeu de données original
df = pd.read_csv("Salary_Data.csv")
```

```python
print(df.columns)
```

```python
print(df.head)
```

# **Mesure de tendance centrale** #

```python
# Moyenne
mean_age = df['Age'].mean()
mean_salary = df['Salary'].mean()

print(f"Moyenne d'âge : {mean_age:.0f} ans")
print(f"Moyenne de salaire : {mean_salary:.2f} €")
```

```python
# Mediane 
median_age = df['Age'].median()
median_salary = df['Salary'].median()

print(f"Médiane d'âge : {median_age} ans")
print(f"Médiane de salaire : {median_salary} €")
```

```python
# Mode 
mode_age = df['Age'].mode()[0]
mode_salary = df['Salary'].mode()[0]

print(f"Mode d'âge : {mode_age} ans")
print(f"Mode de salaire : {mode_salary} €")
```

```python
# Histogramme des salaires
plt.figure(figsize=(10, 6))
sns.histplot(df["Salary"], bins=15, kde=True, color="skyblue", edgecolor="black")

# Moyenne
plt.axvline(df["Salary"].mean(), color='red', linestyle='--', label='Moyenne')
# Médiane
plt.axvline(df["Salary"].median(), color='green', linestyle='-.', label='Médiane')
# Mode
plt.axvline(df["Salary"].mode()[0], color='orange', linestyle=':', label='Mode')

plt.title("Distribution des Salaires avec Moyenne, Médiane et Mode")
plt.xlabel("Salaire (€)")
plt.ylabel("Nombre d'individus")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Histogramme des âges
plt.figure(figsize=(10, 6))
sns.histplot(df["Age"], bins=10, kde=True, color="salmon", edgecolor="black")

plt.axvline(df["Age"].mean(), color='red', linestyle='--', label='Moyenne')
plt.axvline(df["Age"].median(), color='green', linestyle='-.', label='Médiane')
plt.axvline(df["Age"].mode()[0], color='orange', linestyle=':', label='Mode')

plt.title("Distribution des Âges avec Moyenne, Médiane et Mode")
plt.xlabel("Âge")
plt.ylabel("Fréquence")
plt.legend()
plt.grid(True)
plt.show()
```

# **Mesure de dispersion** #

```python
# Etendue
range_age = df["Age"].max() - df["Age"].min()
range_salary = df["Salary"].max() - df["Salary"].min()

print(f"Étendue de l'âge : {range_age} ans")
print(f"Étendue du salaire : {range_salary} €")
```

```python
# Ecart type
std_age = df["Age"].std()
std_salary = df["Salary"].std()

print(f"Écart-type de l'âge : {std_age:.2f}")
print(f"Écart-type du salaire : {std_salary:.2f} €")
```

```python
# Variance
var_age = df["Age"].var()
var_salary = df["Salary"].var()

print(f"Variance de l'âge : {var_age:.2f}")
print(f"Variance du salaire : {var_salary:.2f} €²")
```

```python
# Coefficient de variation
cv_age = std_age / df["Age"].mean()
cv_salary = std_salary / df["Salary"].mean()

print(f"Coefficient de variation (âge) : {cv_age:.2%}")
print(f"Coefficient de variation (salaire) : {cv_salary:.2%}")
```

```python
# Quartiles
quartiles_age = df["Age"].quantile([0.25, 0.5, 0.75])
quartiles_salary = df["Salary"].quantile([0.25, 0.5, 0.75])

print("\nQuartiles de l'âge :")
print(quartiles_age)

print("\nQuartiles du salaire :")
print(quartiles_salary)
```

```python
# Boxplot pour le salaire
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Salary"], color="skyblue")
plt.title("Boxplot du Salaire")
plt.xlabel("Salaire (€)")
plt.grid(True)
plt.show()
```

```python
# Boxplot pour l'âge
plt.figure(figsize=(8, 4))
sns.boxplot(x=df["Age"], color="lightgreen")
plt.title("Boxplot de l'Âge")
plt.xlabel("Âge")
plt.grid(True)
plt.show()
```
